```python
import pandas as pd
import numpy as np
import scipy.stats as stats
from IPython.display import display, Markdown

df_psy = pd.read_excel('Психологи.xlsx')
df_psy.columns = [str(c).strip() for c in df_psy.columns]
display(df_psy.head())
```

Гипотезы (для каждого показателя)
H0: гендерных различий нет среднее значение показателя у юношей и девушек одинаково.
H1: гендерные различия есть средние значения различаются.
Критерий: двухвыборочный t-критерий Стьюдента для независимых выборок (scipy.stats.ttest_ind, equal_var=False — поправка Уэлча). α = 0.05: если p < 0.05 — H0 отвергается (различия статистически значимы).

```python
df_male = df_psy[df_psy['пол'] == 'м']
df_female = df_psy[df_psy['пол'] == 'ж']
print(f"Юноши (м): {len(df_male)}")
print(f"Девушки (ж): {len(df_female)}")
```

```python
results = []
alpha = 0.05

numeric_cols = df_psy.select_dtypes(include=np.number).columns
cols_to_test = [c for c in numeric_cols if c not in ['учащиеся', 'профиль ВУЗа']]

for col in cols_to_test:
    group_m = df_male[col].dropna()
    group_f = df_female[col].dropna()

    t_stat, p_val = stats.ttest_ind(group_m, group_f, equal_var=False)

    conclusion = "Различия есть (но отвергается)" if p_val < alpha else "Различий нет (но не отвергается)"

    results.append({
        'Показатель': col,
        'Ср.знач (м)': round(group_m.mean(), 2),
        'Ср.знач (ж)': round(group_f.mean(), 2),
        't-статистика': round(t_stat, 3),
        'p-value': round(p_val, 4),
        'Вывод (α=0.05)': conclusion
    })

results_df = pd.DataFrame(results)
display(results_df)
```

сильнее всего гендерные различия проявились в личностных и вербальных показателях, у девушек в среднем выше нейротизм, а также результаты по шкалам «пропущенные слова» и «заучивание слов», и именно у этих показателей p-value минимальное, то есть различия самые надёжные. Слабее всего различия выражены в математических и общих интеллектуальных показателях — по шкалам «арифметика», «числовые ряды», «умозаключения» и по «среднему IQ» p-value больше 0.05, значит значимых различий между юношами и девушками там нет
