Задание 2.2. Процентили и SLA

1. Рассчитайте:
◦ среднее значение;
◦ стандартное отклонение;
◦ медиану.
2. Рассчитайте следующие процентили:
p50
p90
p95
p99
используя np.quantile().

```python
np.random.seed(42)
delivery_simulated = np.random.normal(loc=30, scale=5, size=1000)

mean_sim = delivery_simulated.mean()
std_sim = delivery_simulated.std()
median_sim = np.median(delivery_simulated)

print(mean_sim)
print(std_sim)
print(median_sim)

p50, p90, p95, p99 = np.quantile(delivery_simulated, [0.50, 0.90, 0.95, 0.99])
print(p50)
print(p90)
print(p95)
print(p99)

print(p90 <= 40)
```

SlA выполняется, 90 - времени доставки меньше 40 минут, 95 и 99 больше чем 40, значит в срок не попадают. То есть если контролировать и 95, то тогда сроки можно уместить в большинство

Задание 3.1. Bootstrap и доверительный интервал

1. Рассчитайте среднее значение исходной выборки.
2. Реализуйте Bootstrap с количеством итераций:
B = 5000
На каждой итерации:
• сформируйте выборку такого же размера, как x;
• используйте выборку с возвращением;
• рассчитайте её среднее;
• сохраните результат.
Для формирования Bootstrap-выборки используйте:
np.random.choice()
3. Получите центральный 95%-й доверительный интервал для среднего с помощью:
np.quantile()
Используйте квантили:
0.025
0.975
4. Выведите полученный доверительный интервал.

```python
x = np.array([28, 31, 29, 45, 32, 30])
print(x.mean())

B = 5000
boot_means = np.empty(B)

np.random.seed(42)
for i in range(B):
    sample = np.random.choice(x, size=x.size, replace=True)
    boot_means[i] = sample.mean()

ci_low, ci_high = np.quantile(boot_means, [0.025, 0.975])
print(ci_low)
print(ci_high)
```

Задание 3.2. Формулировка гипотез и ошибки I и II рода

1. Сформулируйте нулевую гипотезу H_0
2. Сформулируйте альтернативную гипотезу H_1.
3. Объясните применительно к этому эксперименту:
◦ что означает ошибка I рода;
◦ что означает ошибка II рода.
4. Укажите, какая ошибка соответствует:
◦ ложной тревоге;
◦ пропущенному эффекту.

Нулевая гипотеза H₀: \mu_A = \mu_B — новый алгоритм не меняет среднее время доставки. Альтернативная гипотеза H₁: \mu_B < \mu_A — новый алгоритм уменьшает среднее время доставки. Ошибка I рода — отвергнуть H₀, когда она верна, то есть решить, что алгоритм улучшил доставку, хотя эффекта нет. Ошибка II рода — не отвергнуть H₀, когда она неверна, то есть не заметить реальное улучшение. Ложная тревога соответствует ошибке I рода, пропущенный эффект — ошибке II рода. В ситуации с эксплуатацией за 5 млн руб./год: если алгоритм не улучшает доставку, а мы его внедрили (ошибка I рода), деньги потрачены зря. Значит, к прямым напрасным расходам приводит ошибка I рода.
