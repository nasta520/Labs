```python
np.random.seed(123)
A = np.random.randint(1, 11, size=(3, 3))
print(A)
```

```python
print(np.sum(A[A % 2 == 0]))
```

```python
B = np.dot(np.dot(A, A.T), np.linalg.inv(np.dot(A, A.T)))
print(B)
```

```python
print(np.linalg.det(B))
```

```python
M = np.zeros((3, 3))
for i in range(3):
    for j in range(3):
        sub = np.delete(np.delete(A, i, axis=0), j, axis=1)
        M[i, j] = np.linalg.det(sub)
print(M)
```
